```python
import numpy as np
import matplotlib.pyplot as plt

C = 420
R = 1.0
D_min = 4.0

N = np.array([3, 4, 7, 9, 12])

reuse_ratio = np.sqrt(3 * N)
D = R * reuse_ratio
channels = C / N

print(" N | D/R | D (km) | Channels/Cell")
print("-" * 36)

for n, ratio, d, ch in zip(N, reuse_ratio, D, channels):
    print(f"{n:2d} | {ratio:4.2f} | {d:6.2f} | {ch:11.1f}")

# Find the smallest cluster size that satisfies the minimum distance
valid = N[D >= D_min]

if len(valid) > 0:
    N_opt = min(valid)
    i = np.where(N == N_opt)[0][0]

    print(f"\nMinimum required distance = {D_min} km")
    print(f"Selected cluster size = {N_opt}")
    print(f"Reuse distance = {D[i]:.2f} km")
    print(f"Reuse ratio = {reuse_ratio[i]:.2f}")
    print(f"Channels per cell = {channels[i]:.1f}")
else:
    N_opt = None
    print("\nNo cluster size satisfies the minimum distance.")

# Reuse distance plot
plt.figure()
plt.plot(N, D, "o-")
plt.axhline(D_min, linestyle="--", color="red")

if N_opt is not None:
    plt.plot(N_opt, D[i], "s", color="green", markersize=8)

plt.xlabel("Cluster Size (N)")
plt.ylabel("Reuse Distance (km)")
plt.title("Reuse Distance vs Cluster Size")
plt.grid()
plt.show()

# Channels per cell plot
plt.figure()
plt.plot(N, channels, "o-", color="orange")

if N_opt is not None:
    plt.plot(N_opt, channels[i], "s", color="green", markersize=8)

plt.xlabel("Cluster Size (N)")
plt.ylabel("Channels per Cell")
plt.title("Channels per Cell vs Cluster Size")
plt.grid()
plt.show()
```

```python
import math

TOTAL_CHANNELS = 420
R = 1.0          # cell radius (km)
D_MIN = 4.0      # minimum reuse distance (km)
N_VALUES = [3, 4, 7, 9, 12]
PATH_LOSS_EXP = 4   # path-loss exponent n
K_INTERFERERS = 6   # first-tier co-channel cells

def reuse_distance(N, R):
    return R * math.sqrt(3 * N)

def sir_db(N, n=PATH_LOSS_EXP, k=K_INTERFERERS):
    q = math.sqrt(3 * N)               # co-channel reuse ratio D/R
    return 10 * math.log10((q ** n) / k)

print(f"{'N':>3} {'D (km)':>8} {'Feasible':>9} {'Ch/cell':>9} {'SIR (dB)':>9}")
print("-" * 44)

feasible = []
for N in N_VALUES:
    D = reuse_distance(N, R)
    ok = D >= D_MIN
    ch = TOTAL_CHANNELS / N
    if ok:
        feasible.append(N)
    print(f"{N:>3} {D:>8.3f} {str(ok):>9} {ch:>9.2f} {sir_db(N):>9.2f}")

best = min(feasible)
print(f"\nSmallest feasible N = {best}")
print(f"Channels per cell   = {TOTAL_CHANNELS // best} "
      f"(exact: {TOTAL_CHANNELS / best:.2f})")
print(f"Reuse distance      = {reuse_distance(best, R):.3f} km")
```
